```python
import mne
import numpy as np
from scipy.signal import welch, coherence
from itertools import combinations

# ======================
# 1️⃣ Load and preprocess
# ======================
fname_main = "/Users/roman/PycharmProjects/brain_data/data/15.10.24/2/chistova_alena/main.fif"
fname_rest = "/Users/roman/PycharmProjects/brain_data/data/15.10.24/2/chistova_alena/resting.fif"

raw_main = mne.io.read_raw_fif(fname_main, preload=True)
raw_rest = mne.io.read_raw_fif(fname_rest, preload=True)

raw_main.filter(1., 40.)
raw_rest.filter(1., 40.)

# Pick magnetometers
picks_main = mne.pick_types(raw_main.info, meg='mag')
picks_rest = mne.pick_types(raw_rest.info, meg='mag')

ch_names_main = [raw_main.ch_names[i] for i in picks_main]
ch_names_rest = [raw_rest.info['ch_names'][i] for i in picks_rest]

common_mags = list(set(ch_names_main).intersection(set(ch_names_rest)))
assert len(common_mags) > 0, "No common mag channels found."
print("Using channels:", common_mags)

data_main = raw_main.copy().pick(common_mags).get_data()
data_rest = raw_rest.copy().pick(common_mags).get_data()
sfreq = raw_main.info['sfreq']
```

# window gen

```python
# ======================
# Parameters
# ======================
WINDOW_SEC = 2
STRIDE_SEC = 1
TEST_FRAC = 0.2

window_samples = int(WINDOW_SEC * sfreq)
stride_samples = int(STRIDE_SEC * sfreq)

# ======================
# Function: create full windows only
# ======================
def create_nonoverlapping_windows(data, window_samples):
    """Return only full windows, discarding leftover samples at the end."""
    n_channels, n_times = data.shape
    n_windows = n_times // window_samples  # full windows only
    windows = np.array([
        data[:, i*window_samples:(i+1)*window_samples]
        for i in range(n_windows)
    ])
    return windows

# ======================
# Function: split data into train/test first, then create windows
# ======================
def split_and_window(data, test_frac=0.2):
    n_samples = data.shape[1]
    split_idx = int(n_samples * (1 - test_frac))

    # Split in time
    train_data = data[:, :split_idx]
    test_data  = data[:, split_idx:]

    # Create full windows separately
    train_windows = create_nonoverlapping_windows(train_data, window_samples)
    test_windows  = create_nonoverlapping_windows(test_data, window_samples)

    return train_windows, test_windows

# ======================
# Apply to each condition
# ======================
wm_train, wm_test = split_and_window(data_main, TEST_FRAC)
wr_train, wr_test = split_and_window(data_rest, TEST_FRAC)

# ======================
# Combine and label
# ======================
X_train = np.concatenate([wm_train, wr_train], axis=0)
y_train = np.concatenate([np.ones(len(wm_train)), np.zeros(len(wr_train))])

X_test = np.concatenate([wm_test, wr_test], axis=0)
y_test = np.concatenate([np.ones(len(wm_test)), np.zeros(len(wr_test))])

print("Train windows:", X_train.shape, "Test windows:", X_test.shape)
```

```python
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

CHANNEL = "MEG2511"

# Find channel index
ch_idx = common_mags.index(CHANNEL)

# window 0 is a bit broken, take second
win_main = wm_train[0, ch_idx]   # thinking
win_rest = wr_train[1, ch_idx]   # resting

t = np.arange(len(win_main)) / sfreq
```

```python
def bandpass(data, low, high, fs, order=4):
    nyq = 0.5 * fs
    b, a = butter(order, [low/nyq, high/nyq], btype='band')
    return filtfilt(b, a, data)
```

```python
fig, axes = plt.subplots(len(bands)+1, 2, figsize=(12, 10), sharex=True)

# ---- Raw signal ----
axes[0, 0].plot(t, win_main)
axes[0, 0].set_title("Thinking - Raw")

axes[0, 1].plot(t, win_rest)
axes[0, 1].set_title("Resting - Raw")

# ---- Bandpass signals ----
for i, (band_name, (low, high)) in enumerate(bands.items(), start=1):
    main_filt = bandpass(win_main, low, high, sfreq)
    rest_filt = bandpass(win_rest, low, high, sfreq)

    axes[i, 0].plot(t, main_filt)
    axes[i, 0].set_title(f"Thinking - {band_name}")

    axes[i, 1].plot(t, rest_filt)
    axes[i, 1].set_title(f"Resting - {band_name}")

plt.tight_layout()
plt.show()
```
